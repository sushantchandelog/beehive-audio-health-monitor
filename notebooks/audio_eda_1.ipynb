```python
# importing the Independencies 

import librosa 
import librosa.display
import matplotlib.pyplot as plt
import numpy as np
import IPython as ipd
import os


plt.style.use('ggplot')

print('import Done')
```

```python
raw_data_dr = "../data/raw"
processed_data_dr = "data/processed/five_features.csv"

print(raw_data_dr)
```

```python
#labelling the Data
# classes 1mean healthya and 0 means distressed (no Queen)
import gc
classes = {'QueenBee Present':1, 'QueenBee Absent':0}

features =[]
labels = []


for f, l in classes.items():
    folder_path = os.path.join(raw_data_dr , f)
    print(f'Processing {f} audio files')
    
    count = 0
    for file_name in os.listdir(folder_path):
        if file_name.endswith('.wav'):
            file_path = os.path.join(folder_path , file_name)
            
            # loading the Audio files 
            y , sr = librosa.load(file_path , sr=22050)
            
            #extracting 40 MFCC
            mfcc = librosa.feature.mfcc(y=y , sr=sr , n_mfcc = 40)
            
            #flatteing the 2d array to 1 d by taking the mean
            mfccs_mean = np.mean(mfcc.T , axis=0)
            
            # appending the Empty lists
            features.append(mfccs_mean)
            labels.append(labels)
            
            del y , mfcc , mfccs_mean 
    gc.collect()
```
